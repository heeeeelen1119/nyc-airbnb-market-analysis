# NYC Airbnb exploratory analysis

This notebook loads the processed data and reproduces a small set of descriptive checks. Run the cleaning pipeline first.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
listings = pd.read_parquet(root / 'data/processed/listings_clean.parquet')
listings.groupby(['borough', 'room_type']).price.agg(['size', 'mean', 'median']).round(2)
```

```python
# Pandas counterpart to a SQL median cross-check
listings.groupby('room_type').price.median().sort_values(ascending=False)
```
